## Which campaign type generates the highest revenue and ROI?

```python
query = """
SELECT
    campaign_type,
    SUM(revenue) AS Total_Revenue,
    ROUND(AVG(roi),2) AS Avg_ROI
FROM campaigns
GROUP BY campaign_type
ORDER BY Total_Revenue DESC;
"""

result = pd.read_sql(query, conn)

result
```

## What percentage of total revenue comes from each audience segment?

```python
query = """
SELECT
    target_audience,
    SUM(revenue) AS Revenue,
    ROUND(
        100.0 * SUM(revenue) /
        (SELECT SUM(revenue) FROM campaigns),
        2
    ) AS Revenue_Share
FROM campaigns
GROUP BY target_audience
ORDER BY Revenue DESC;
"""

result = pd.read_sql(query, conn)

result
```

## Which campaigns have negative ROI and how much revenue are they collectively wasting?

```python
query = """
SELECT
    campaign_type,
    target_audience,
    COUNT(*) AS Campaigns,
    SUM(revenue) AS Revenue,
    SUM(acquisition_cost) AS Cost
FROM campaigns
WHERE roi < 0
GROUP BY
    campaign_type,
    target_audience
ORDER BY Cost DESC;
"""

result = pd.read_sql(query, conn)

result
```

## For each campaign type, what is the revenue earned per rupee spent on acquisition?

```python
query = """
SELECT
    campaign_type,
    ROUND(
        SUM(revenue) /
        SUM(acquisition_cost),
        2
    ) AS Revenue_Per_Rupee
FROM campaigns
GROUP BY campaign_type
ORDER BY Revenue_Per_Rupee DESC;
"""

result = pd.read_sql(query, conn)

result
```

## What does the full conversion funnel look like for each audience segment?

```python
query = """
SELECT

    target_audience,

    SUM(impressions) AS Impressions,

    SUM(clicks) AS Clicks,

    SUM(leads) AS Leads,

    SUM(conversions) AS Conversions,

    ROUND(
        100.0 * SUM(clicks) /
        SUM(impressions),
        2
    ) AS CTR,

    ROUND(
        100.0 * SUM(leads) /
        SUM(clicks),
        2
    ) AS Lead_Rate,

    ROUND(
        100.0 * SUM(conversions) /
        SUM(leads),
        2
    ) AS Conversion_Rate

FROM campaigns

GROUP BY target_audience;
"""

result = pd.read_sql(query, conn)

result
```

## Which audience segment has the highest drop-off between leads and final conversion?

```python
query = """
SELECT

    target_audience,

    ROUND(
        100.0 * SUM(conversions) /
        SUM(leads),
        2
    ) AS Lead_To_Conversion_Rate

FROM campaigns

GROUP BY target_audience

ORDER BY Lead_To_Conversion_Rate;
"""

result = pd.read_sql(query, conn)

result
```


## During which months does ROI drop the most, and what campaign types are running then?

```python
query = """
WITH monthly_roi AS (
    SELECT
        Month_Year,
        ROUND(AVG(ROI), 2) AS Avg_ROI
    FROM campaigns
    GROUP BY Month_Year
),

roi_change AS (
    SELECT
        Month_Year,
        Avg_ROI,
        LAG(Avg_ROI) OVER (ORDER BY Month_Year) AS Previous_Month_ROI,
        ROUND(
            Avg_ROI - LAG(Avg_ROI) OVER (ORDER BY Month_Year),
            2
        ) AS ROI_Change
    FROM monthly_roi
),

campaign_breakdown AS (
    SELECT
        Month_Year,
        Campaign_Type,
        COUNT(*) AS Campaign_Count,
        ROUND(AVG(ROI), 2) AS Campaign_Avg_ROI
    FROM campaigns
    GROUP BY Month_Year, Campaign_Type
)

SELECT
    r.Month_Year,
    r.Avg_ROI,
    r.Previous_Month_ROI,
    r.ROI_Change,
    c.Campaign_Type,
    c.Campaign_Count,
    c.Campaign_Avg_ROI
FROM roi_change r
JOIN campaign_breakdown c
    ON r.Month_Year = c.Month_Year
WHERE r.ROI_Change < 0
ORDER BY r.ROI_Change ASC, c.Campaign_Avg_ROI ASC;
"""

result = pd.read_sql(query, conn)
result
```

## Do Hindi campaigns outperform English campaigns for Tier 2 City customers?

```python
query = """
SELECT

language,

SUM(revenue) AS Revenue,

ROUND(
AVG(roi),
2
) AS Avg_ROI,

ROUND(
AVG(engagement_score),
2
) AS Avg_Engagement

FROM campaigns

WHERE target_audience = 'Tier 2 City Customers'

GROUP BY language;
"""

result = pd.read_sql(query, conn)

result
```

## Which language has the highest engagement score and does it also convert better?

```python
query = """
SELECT

language,

ROUND(
AVG(engagement_score),
2
) AS Engagement,

ROUND(
100.0 * SUM(conversions) /
SUM(leads),
2
) AS Conversion_Rate

FROM campaigns

GROUP BY language;
"""

result = pd.read_sql(query, conn)

result
```

## Which combination of campaign type, audience, and language gives the highest ROI?

```python
query = """
SELECT *

FROM (

SELECT

campaign_type,

target_audience,

language,

ROUND(
AVG(roi),
2
) AS Avg_ROI,

RANK() OVER(

ORDER BY AVG(roi) DESC

) AS Ranking

FROM campaigns

GROUP BY

campaign_type,

target_audience,

language

)

WHERE Ranking <= 10;
"""

result = pd.read_sql(query, conn)

result
```

## Within each campaign type, which audience segment delivers the best conversion rate?

```python
query = """
WITH rates AS (

SELECT

campaign_type,

target_audience,

ROUND(
100.0 * SUM(conversions) /
SUM(leads),
2
) AS Conversion_Rate

FROM campaigns

GROUP BY

campaign_type,

target_audience

)

SELECT *

FROM (

SELECT *,

RANK() OVER(

PARTITION BY campaign_type

ORDER BY Conversion_Rate DESC

) AS Rank_No

FROM rates

)

WHERE Rank_No = 1;
"""

result = pd.read_sql(query, conn)

result
```

## Campaigns Performing Above Average ROI

```python
query = """
WITH avg_roi AS (
    SELECT
        campaign_type,
        AVG(roi) AS Avg_ROI
    FROM campaigns
    GROUP BY campaign_type
),

ranked_campaigns AS (
    SELECT
        c.campaign_id,
        c.campaign_type,
        c.roi,
        a.Avg_ROI,

        ROW_NUMBER() OVER (
            PARTITION BY c.campaign_type
            ORDER BY c.roi DESC
        ) AS rn

    FROM campaigns c

    JOIN avg_roi a
        ON c.campaign_type = a.campaign_type

    WHERE c.roi > a.Avg_ROI
)

SELECT
    campaign_id,
    campaign_type,
    roi,
    ROUND(Avg_ROI, 2) AS Avg_ROI
FROM ranked_campaigns
WHERE rn <= 5
ORDER BY campaign_type, roi DESC;
"""

result = pd.read_sql(query, conn)

result
```

## Cumulative Revenue

```python
query = """
WITH monthly AS (
    SELECT
        Month_Year,
        SUM(Revenue) AS Revenue
    FROM campaigns
    GROUP BY Month_Year
)

SELECT
    Month_Year,
    Revenue,
    SUM(Revenue) OVER (
        ORDER BY Month_Year
    ) AS Cumulative_Revenue
FROM monthly
ORDER BY Month_Year;
"""
result = pd.read_sql(query, conn)

result
```
